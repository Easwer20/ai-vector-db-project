```python
import psycopg2
import numpy as np
from langchain_huggingface import HuggingFaceEmbeddings
```

```python
texts = [
    "Artificial Intelligence is transforming healthcare through predictive analytics.",
    
    "PostgreSQL with pgvector is commonly used for storing vector embeddings in RAG systems.",
    
    "LangChain helps developers build applications powered by large language models.",
    
    "Machine learning models learn patterns from historical data.",
    
    "Python is widely used in AI engineering because of its rich ecosystem.",
    
    "Retrieval Augmented Generation combines search with language models.",
    
    "Embeddings convert text into numerical vector representations.",
    
    "OpenAI provides APIs for embeddings, chat models, and image generation.",
    
    "Vector databases are optimized for similarity search operations.",
    
    "Jupyter Notebook is useful for data science and AI experimentation."
]
```

```python
embeddings = HuggingFaceEmbeddings(model_name="sentence-transformers/all-mpnet-base-v2")
embedding_lst=[]

for text in texts:
    embedding_lst.append(embeddings.embed_query(text))
```

```python
conn=psycopg2.connect("dbname=DBname user=user")
curr=conn.cursor()
for i in range(len(embedding_lst)):
    embedding=embedding_lst[i]
    content=texts[i]
    curr.execute("INSERT INTO items(content,embedding) VALUES(%s, %s)",(content,embedding))
conn.commit()
curr.close()
conn.close()
```

```python
new_text ="Deep learning models use neural networks with multiple layers to solve complex AI tasks."
new_embedding = embeddings.embed_query(new_text)
```

```python
conn=psycopg2.connect("dbname=DBname user=user")
curr=conn.cursor()
curr.execute("""SELECT id,content FROM items ORDER BY embedding <-> %s::vector LIMIT 5""",(new_embedding,))
```
